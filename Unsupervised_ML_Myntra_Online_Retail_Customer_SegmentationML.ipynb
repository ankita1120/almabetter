## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Handle missing values by dropping rows with missing CustomerID or Description
df.dropna(subset=['Description', 'CustomerID'], inplace=True)

# Convert InvoiceDate to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Remove rows with negative Quantity or UnitPrice
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Display the first few rows of the cleaned data
display(df.head())

# Check the shape of the cleaned data
print("\nShape after cleaning:", df.shape)

# Check for remaining missing values
print("\nMissing values after cleaning:\n", df.isnull().sum())

# Check for remaining duplicate values
print("\nDuplicate values after cleaning:", df.duplicated().sum())
````

### What all manipulations have you done and insights you found?

Handled Missing Values: Rows with missing Description or CustomerID were dropped. This was necessary because CustomerID is crucial for customer segmentation and Description is important for understanding the products.
Converted InvoiceDate to Datetime: The InvoiceDate column was converted from an object type to a datetime object. This is essential for any time-series analysis or extracting time-based features like month, year, or day.
Removed Rows with Negative Quantity or UnitPrice: Rows where Quantity or UnitPrice were less than or equal to zero were removed. Negative quantities usually indicate returns, which might be analyzed separately or excluded depending on the objective. Zero or negative unit prices are likely data errors.
Removed Duplicate Rows: Identified and removed duplicate rows to ensure each transaction record is unique.
Insights Found:

Reduced Dataset Size: After these manipulations, the dataset size significantly reduced from 541,909 rows to 392,692 rows. This indicates a substantial amount of data was either missing required information (CustomerID) or contained invalid values (negative quantity/price, duplicates).
Clean Data for Analysis: The cleaned dataset now has no missing values in the critical Description and CustomerID columns, no negative quantities or unit prices, and no duplicate rows, making it ready for further analysis and visualization.
CustomerID Missingness: The large number of rows dropped due to missing CustomerID (over 135,000) highlights that a significant portion of transactions were made by customers who were not identified. This could be important to consider for customer-centric analyses.
These steps have resulted in a cleaner and more reliable dataset for exploring purchasing trends, product performance, and customer behavior.

#### Chart - 1 Total Sales Over Time

````python
# Chart - 1 Total Sales Over Time

# Calculate total price for each transaction
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']

# Extract month and year from InvoiceDate
df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')

# Group by month and calculate total sales
monthly_sales = df.groupby('InvoiceMonth')['TotalPrice'].sum().reset_index()

# Convert InvoiceMonth back to datetime for plotting
monthly_sales['InvoiceMonth'] = monthly_sales['InvoiceMonth'].dt.to_timestamp()

# Plotting the total sales over time
plt.figure(figsize=(12, 6))
sns.lineplot(data=monthly_sales, x='InvoiceMonth', y='TotalPrice')
plt.title('Total Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

 chose a line plot to visualize the total sales over time because line plots are ideal for showing trends and changes in a variable over a continuous period. In this case, it allows us to easily see how the total sales have evolved month by month.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
from scipy import stats

z = np.abs(stats.zscore(df['TotalPrice']))
threshold = 3
df_no_outliers = df[(z < threshold)]

Q1 = df_no_outliers['TotalPrice'].quantile(0.25)
Q2 = df_no_outliers['TotalPrice'].quantile(0.50)
Q3 = df_no_outliers['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Option 1: Remove outliers
df_no_outliers = df_no_outliers[(df_no_outliers['TotalPrice'] >= lower_bound) & (df_no_outliers['TotalPrice'] <= upper_bound)]

# Option 2: Winsorization
df_no_outliers['TotalPrice'] = np.where(df_no_outliers['TotalPrice'] < lower_bound, lower_bound, df_no_outliers['TotalPrice'])

print("Before removing outliers:", df.shape)
print("After removing outliers:", df_no_outliers.shape)

````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Z-score: You removed rows where the absolute Z-score of TotalPrice was greater than 3. This method assumes the data is normally distributed and identifies outliers as data points that are a certain number of standard deviations away from the mean.
Interquartile Range (IQR): After applying the Z-score method, you further removed rows where TotalPrice was below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR. This is a common non-parametric method for outlier detection that is less sensitive to the distribution of the data.
These techniques were likely used to remove extreme values in the TotalPrice column that could skew analysis and potentially negatively impact machine learning models.

````python
import re

contraction_mapping = {
    "can't": "can not",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'d": " would",
    "'ll": " will",
    "'t": " not",
    "'ve": " have",
    "'m": " am"
}

def expand_contractions_manual(text):
    # Replace each contraction found in the mapping
    for contraction, expansion in contraction_mapping.items():
        text = re.sub(contraction, expansion, text)
    return text

text = "She won't go because she's tired."
expanded_text = expand_contractions_manual(text)
print(expanded_text)
# Output: "She will not go because she is tired."
````

````python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import nltk

# Download necessary NLTK data
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('punkt_tab')
nltk.download('omw-1.4') # Add this line to download the missing resource


text = "The striped bats are hanging on their feet for best."

# Lowercase
text = text.lower()

# Tokenize
tokens = word_tokenize(text)

# Remove stopwords
tokens = [word for word in tokens if word.isalpha() and word not in stopwords.words('english')]

# Stemming
ps = PorterStemmer()
stemmed_tokens = [ps.stem(word) for word in tokens]

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in tokens]

print("Stemmed:", stemmed_tokens)
print("Lemmatized:", lemmatized_tokens)
````

Why it was used: TF-IDF is often used because it gives more weight to words that are important within a document but not very common across the entire corpus. This helps to highlight terms that are more specific and potentially more informative for distinguishing between documents compared to very common words (even after removing stop words).
Both techniques are used to convert the textual 'Description' data into numerical feature vectors, which are required as input for most machine learning algorithms you might use for tasks like product categorization or clustering based on descriptions.

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np # Import numpy for NaN

# Example: testing k from 2 to 10
range_n_clusters = list(range(2, 11))
silhouette_scores = []
inertias = []

# Assuming X is your feature matrix defined in previous steps (e.g., scaled numerical features)
# Make sure X is defined before running this cell.

for n_clusters in range_n_clusters:
    print(f"Attempting KMeans with k = {n_clusters}")
    # Initialize KMeans with the current number of clusters
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10) # Added n_init to suppress warning

    # Fit KMeans and get cluster labels
    cluster_labels = kmeans.fit_predict(X)

    # Check the number of unique labels formed
    n_unique_labels = len(set(cluster_labels))
    print(f"  Number of unique labels formed: {n_unique_labels}")

    # Get Inertia (Within-cluster sum of squares)
    inertias.append(kmeans.inertia_)

    # Initialize silhouette_avg to NaN
    silhouette_avg = np.nan

    # Calculate Silhouette Score only if more than 1 cluster is formed
    if n_unique_labels > 1 and n_unique_labels < len(cluster_labels): # Also ensure n_labels < n_samples
        # Calculating Silhouette Score on large datasets can be slow.
        # Using sample_size calculates the score on a random subset of the data.
        # Adjust sample_size based on your dataset size and computational resources.
        try:
            if len(cluster_labels) > 10000: # Ensure enough samples for calculation
                # Ensure sample_size is not greater than the number of data points
                sample_size = min(10000, len(cluster_labels) -1) # Ensure sample_size is valid
                silhouette_avg = silhouette_score(X, cluster_labels, sample_size=sample_size) # Use a sample for faster calculation
            else:
                silhouette_avg = silhouette_score(X, cluster_labels) # Calculate on full data if small enough
        except ValueError as e:
             print(f"  Could not calculate Silhouette Score for k={n_clusters}: {e}")
             silhouette_avg = np.nan # Ensure it's nan if calculation fails

    else:
        print(f"  Skipping Silhouette Score calculation for k={n_clusters} as only 1 cluster was formed or n_labels >= n_samples.")

    silhouette_scores.append(silhouette_avg)


# ----------------- Plot Silhouette Score vs. k -----------------
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
# Plot silhouette scores, handling potential NaNs
plt.plot(range_n_clusters, silhouette_scores, marker='o', color='teal')
plt.title('Silhouette Score vs. Number of Clusters')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(range_n_clusters) # Ensure all tested k values are shown on x-axis
plt.grid(True)

# ----------------- Plot Inertia vs. k (Elbow Method) -----------------
plt.subplot(1, 2, 2)
plt.plot(range_n_clusters, inertias, marker='o', color='purple')
plt.title('Inertia (SSE) vs. Number of Clusters (Elbow Method)')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia (SSE)')
plt.xticks(range_n_clusters) # Ensure all tested k values are shown on x-axis
plt.grid(True)

plt.tight_layout()
plt.show()
````

K-Means Clustering:

Explanation: K-Means is an iterative algorithm that partitions the data into k predefined, non-overlapping subgroups (clusters) where every data point belongs to only one group. It works by:
Initializing k centroids (randomly or using a smart initialization method).
Assigning each data point to the nearest centroid based on Euclidean distance.
Recalculating the centroids as the mean of all data points assigned to that cluster.
Repeating steps 2 and 3 until the centroids no longer change significantly or a maximum number of iterations is reached.
Understanding Feature Importance: For K-Means, you don't typically use traditional feature importance tools. Features with larger differences in their mean values across clusters are more important in differentiating those segments.
DBSCAN Clustering:

Explanation: DBSCAN is a density-based clustering algorithm that groups together points that are closely packed together, marking as outliers points that lie alone in low-density regions. It works based on two parameters:
eps: The maximum distance between two samples for one to be considered as in the neighborhood of the other.
min_samples: The number of samples (or total weight) in a neighborhood for a point to be considered as a core point. DBSCAN identifies core points, border points, and noise points based on these parameters and forms clusters from the dense regions.
Understanding Feature Importance: Similar to K-Means, traditional feature importance methods are not directly applicable to DBSCAN. To understand feature importance, you would analyze the distribution of feature values within each cluster (excluding noise, which forms its own category). For unsupervised learning like clustering, the goal is not prediction of a predefined target, but rather discovering patterns and structure in the data. Therefore, these supervised explainability tools are not directly applicable. Understanding "feature importance" in this context is about identifying which features were most influential in grouping the data points into the discovered clusters.
